## 2️⃣ 분석 결과 탐색

```python
from pathlib import Path

# 가장 최신의 출력 폴더 찾기
output_dir = Path('../outputs')
latest_output = sorted(output_dir.glob('*/'))[-1] if output_dir.exists() else None

if latest_output:
    print(f'최신 출력 폴더: {latest_output.name}')
    print(f'\n📊 생성된 파일들:')
    for file in sorted(latest_output.glob('*')):
        size_mb = file.stat().st_size / (1024*1024)
        print(f'  - {file.name} ({size_mb:.1f}MB)' if size_mb >= 1 else f'  - {file.name}')
else:
    print('출력 폴더가 없습니다. preprocess.py를 먼저 실행하세요.')
```

## 🔍 CSV 통계 확인

```python
import pandas as pd
from pathlib import Path

output_dir = Path('../outputs')
latest_output = sorted(output_dir.glob('*/'))[-1] if output_dir.exists() else None

if latest_output:
    csv_files = list(latest_output.glob('*.csv'))
    print(f'CSV 통계 파일 ({len(csv_files)}개):')
    for csv_file in sorted(csv_files):
        print(f'\n📄 {csv_file.name}')
        df = pd.read_csv(csv_file)
        print(df.to_string())
```

## 🗺️ 지도 이미지 확인

```python
from PIL import Image
from pathlib import Path
import matplotlib.pyplot as plt

output_dir = Path('../outputs')
latest_output = sorted(output_dir.glob('*/'))[-1] if output_dir.exists() else None

if latest_output:
    png_files = sorted(latest_output.glob('*.png'))
    if png_files:
        # 첫 번째 이미지만 표시
        img_path = png_files[0]
        img = Image.open(img_path)
        
        plt.figure(figsize=(12, 8))
        plt.imshow(img)
        plt.axis('off')
        plt.title(f'{img_path.name}', fontsize=12)
        plt.tight_layout()
        plt.show()
        
        print(f'\n총 {len(png_files)}개의 지도 파일:')
        for png_file in png_files:
            print(f'  - {png_file.name}')
```
